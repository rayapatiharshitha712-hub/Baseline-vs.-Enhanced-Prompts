```python
from google.colab import userdata
from groq import Groq
```

```python
api_key = userdata.get("GROQ_API_KEY")
client = Groq(api_key=api_key)
```

```python
# -------------------------------
# Baseline Prompt
# -------------------------------

baseline_prompt = "Write a one-paragraph bio of Ada Lovelace."

baseline_response = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "user", "content": baseline_prompt}
    ]
)

# -------------------------------
# Enhanced Prompt
# -------------------------------

enhanced_prompt = """
You are a history teacher.

Write about Ada Lovelace.

Include:
1. Early life
2. Contribution to computing
3. Legacy

Write in simple English.
Use exactly 4 bullet points.
"""

enhanced_response = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "user", "content": enhanced_prompt}
    ]
)

# -------------------------------
# Display Results
# -------------------------------

print("========== Baseline Prompt ==========\n")
print(baseline_prompt)

print("\n========== Baseline Output ==========\n")
print(baseline_response.choices[0].message.content)

print("\n\n========== Enhanced Prompt ==========\n")
print(enhanced_prompt)

print("\n========== Enhanced Output ==========\n")
print(enhanced_response.choices[0].message.content)
```
